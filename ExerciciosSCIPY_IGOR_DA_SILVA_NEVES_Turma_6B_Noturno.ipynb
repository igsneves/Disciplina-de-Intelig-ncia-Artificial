Exercicio 1

```python
import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)

print("Integral:", resultado)
```

Exercicio 2

```python
import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return x - y

solucao = solve_ivp(f, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("Valores de y:")
print(solucao.y[0])
```

Exercicio 3

```python
import numpy as np
from scipy.optimize import brentq

def f(x):
    return x**4 - 3*x**3 + 2

raiz1 = brentq(f, -2, 0)
raiz2 = brentq(f, 0, 1)
raiz3 = brentq(f, 1, 2)
raiz4 = brentq(f, 2, 4)

print(raiz1)
print(raiz2)
print(raiz3)
print(raiz4)
```

Exercicio 4

```python
from scipy.optimize import minimize_scalar

def f(x):
    return -x**2 + 4*x + 1

resultado = minimize_scalar(lambda x: -f(x))

print("Valor de x:", resultado.x)
print("Valor máximo:", f(resultado.x))
```

```python
import numpy as np
from scipy.fft import fft, fftfreq

t = np.linspace(0, 1, 1000)

sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)

transformada = fft(sinal)
frequencias = fftfreq(len(t), t[1] - t[0])

print("Transformada de Fourier:")
print(transformada)
```

Exercicio 10

Exercicio 12

```python
import numpy as np
from scipy.optimize import root_scalar

def f(x):
    return np.exp(x) - x**2

resultado = root_scalar(f, bracket=[-1, 1])

print("Raiz:", resultado.root)
```

Exercicio 14

```python
from scipy.optimize import minimize

def f(valores):
    x, y = valores
    return (x - 2)**2 + (y - 3)**2

resultado = minimize(f, [0, 0])

print("x =", resultado.x[0])
print("y =", resultado.x[1])
print("Valor mínimo:", resultado.fun)
```

Exercicio 15

```python
import numpy as np
from scipy.fft import fft, ifft

sinal = np.random.rand(10) + 1j * np.random.rand(10)

transformada = fft(sinal)

sinal_original = ifft(transformada)

print("Sinal:")
print(sinal)

print("Transformada inversa:")
print(sinal_original)
```

Exercicio 16

```python
import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return [y[1], -y[0]]

solucao = solve_ivp(
    f,
    [0, 10],
    [1, 0],
    t_eval=np.linspace(0, 10, 100)
)

print(solucao.y[0])
```
