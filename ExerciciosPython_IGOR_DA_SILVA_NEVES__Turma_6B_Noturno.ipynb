Exercicio 21

```python
lista = [1, 3, 5, 7, 9, 11, 13, 15];

numero = int(input("Digite o número que deseja buscar:"));

inicio = 0;
fim = len(lista) -1;

while inicio <= fim:
  meio = (inicio + fim) // 2

  if lista[meio] == numero:
    print(f"O número {numero} foi encontrado na posição {meio}.")
    break
  elif lista[meio] < numero:
    inicio = meio + 1
  else:
    fim = meio - 1
else:
  print("Numero não encontrado")
```

Exercicio 22

```python
mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

resultado = ""

for letra in mensagem:
    if letra.isalpha():
        codigo = ord(letra.lower()) - ord('a')
        novo_codigo = (codigo + deslocamento) % 26
        resultado += chr(novo_codigo + ord('a'))
    else:
        resultado += letra

print(resultado)
```

Exercicio 30

```python
palavra = "python"
tentativas = 6
letras_acertadas = []

while tentativas > 0:
    palavra_atual = ""

    for letra in palavra:
        if letra in letras_acertadas:
            palavra_atual += letra
        else:
            palavra_atual += "_"

    print("Palavra:", palavra_atual)
    print(f"Tentativas restantes: {tentativas}")

    if "_" not in palavra_atual:
        print("Parabens, voce acertou a palavra!")
        break

    letra = input("Digite uma letra: ").lower()

    if letra in palavra:
        print("Voce acertou uma letra!")
        letras_acertadas.append(letra)
    else:
        print("Voce errou!")
        tentativas -= 1

else:
    print(f"Voce perdeu! A palavra era: {palavra}")
```

Exercicio 46

```python
lista = [10, 20, 30, 40, 50]

soma = 0

for n in lista:
    soma += n

media = soma / len(lista)

soma_desvios = 0

for n in lista:
    soma_desvios += (n - media) ** 2

desvio_padrao = (soma_desvios / len(lista)) ** 0.5

print(f"Média: {media}")
print(f"Desvio padrão: {desvio_padrao}")
```

Exercicio 48

```python
lista = [5, 2, 8, 1, 3]

for i in range(1, len(lista)):
    atual = lista[i]
    j = i - 1

    while j >= 0 and lista[j] > atual:
        lista[j + 1] = lista[j]
        j -= 1

    lista[j + 1] = atual

print(lista)
```
