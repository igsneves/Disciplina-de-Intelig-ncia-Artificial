Exercicio 8

```python
import pandas as pd

dados = {
    "Produto": ["Notebook", "Celular", "Mouse", "Teclado", "Monitor"],
    "Categoria": ["Informática", "Celular", "Informática", "Informática", "Informática"],
    "Preço": [2500, 1800, 80, 150, 900],
    "Quantidade": [5, 8, 20, 15, 7]
}

df = pd.DataFrame(dados)

resultado = df.groupby("Categoria")["Quantidade"].sum()

print(resultado)
```

Exercicio 13

```python
import pandas as pd

dados = {
    "Produto": ["Arroz", "Café", "Mouse", "Teclado", "Feijão"],
    "Categoria": ["Alimento", "Alimento", "Informática", "Informática", "Alimento"],
    "Quantidade": [10, 5, 20, 15, 30]
}

df = pd.DataFrame(dados)

tabela = pd.pivot_table(
    df,
    values="Quantidade",
    index="Categoria",
    aggfunc="sum"
)

print(tabela)
```
