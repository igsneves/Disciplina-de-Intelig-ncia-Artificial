```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = x**3

plt.plot(x, y)
plt.title("Gráfico de y = x³")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1, 10, 100)

y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("y = eˣ")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("y = log(x)")

plt.show()
```

Exercicio 8

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

y = np.sin(x)

erro = 0.2

plt.plot(x, y)
plt.fill_between(x, y - erro, y + erro, alpha=0.2)

plt.title("Função seno com intervalo")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

Exercicio 10

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

y1 = x
y2 = x**2
y3 = x**3

plt.plot(x, y1, color="red", linestyle="-")
plt.plot(x, y2, color="blue", linestyle="--")
plt.plot(x, y3, color="green", linestyle=":")

plt.title("Três funções")
plt.show()
```

Exercicio 11

```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C"]

grupo1 = [10, 20, 15]
grupo2 = [15, 10, 20]
grupo3 = [5, 15, 10]

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")

total = [grupo1[i] + grupo2[i] for i in range(3)]

plt.bar(categorias, grupo3, bottom=total, label="Grupo 3")

plt.legend()
plt.title("Barras empilhadas")
plt.show()
```

```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E"]
valores = [20, -15, 30, -10, 25]

cores = ["blue" if valor >= 0 else "red" for valor in valores]

plt.bar(categorias, valores, color=cores)

plt.title("Valores positivos e negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

y1 = x
y2 = x**2

plt.plot(x, y1, label="y = x")
plt.plot(x, y2, label="y = x²")

plt.legend()

plt.title("Duas séries")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```
