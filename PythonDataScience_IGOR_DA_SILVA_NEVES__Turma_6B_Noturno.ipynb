```python
import numpy as np

array1 = np.random.randint(1, 10, 20)
array2 = np.random.randint(1, 10, 20)

produto_escalar = np.dot(array1, array2)

norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)

cos_theta = produto_escalar / (norma1 * norma2)

angulo = np.arccos(cos_theta)

print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo em radianos:", angulo)
print("Ângulo em graus:", np.degrees(angulo))
```

```python
import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return x - y

solucao = solve_ivp(f, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("Valores de y:")
print(solucao.y[0])
```

Exercicio 3

```python
import numpy as np
from scipy.optimize import brentq

def f(x):
    return x**4 - 3*x**3 + 2

raiz1 = brentq(f, -2, 0)
raiz2 = brentq(f, 0, 1)
raiz3 = brentq(f, 1, 2)
raiz4 = brentq(f, 2, 4)

print(raiz1)
print(raiz2)
print(raiz3)
print(raiz4)
```

```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E"]
valores = [20, -15, 30, -10, 25]

cores = ["blue" if valor >= 0 else "red" for valor in valores]

plt.bar(categorias, valores, color=cores)

plt.title("Valores positivos e negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

Exercicio 19

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

y1 = x
y2 = x**2

plt.plot(x, y1, label="y = x")
plt.plot(x, y2, label="y = x²")

plt.legend()

plt.title("Duas séries")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = x**2

plt.plot(x, y)

plt.title("y = x²")
plt.xlabel("x")
plt.ylabel("y")

plt.xticks(np.arange(-10, 11, 2))
plt.yticks(np.arange(0, 101, 10))

plt.grid(True, linestyle="--", linewidth=0.8)

plt.show()
```
